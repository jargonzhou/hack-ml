# Demonstration

```python
# LlamaIndex Framework > Integrations > Llm > Hugging Face LLMs
# https://developers.llamaindex.ai/python/framework/integrations/llm/huggingface/

from llama_index.core import VectorStoreIndex, SimpleDirectoryReader, Settings, set_global_handler
from llama_index.llms.huggingface import HuggingFaceLLM
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
from llama_index.core.callbacks import CallbackManager, LlamaDebugHandler
import torch

# Bind globally BEFORE any index loading
debug_handler = LlamaDebugHandler(print_trace_on_end=True)
Settings.callback_manager = CallbackManager([debug_handler])
set_global_handler("simple")

# model_name = "google/gemma-4-E4B-it"
model_name = "google/gemma-3n-E4B-it"
# embed_model_name = "google/embeddinggemma-300m"
embed_model_name = "D:/software/huggingface/hub/models--google--embeddinggemma-300m/snapshots/57c266a740f537b4dc058e1b0cda161fd15afa75"

llm = HuggingFaceLLM(
    model_name=model_name,
    tokenizer_name=model_name,
    # trust_remote_code=True,
    model_kwargs={
        "dtype": torch.bfloat16,
        # "device_map": "auto",
        # "trust_remote_code": True,
    })
embed_model = HuggingFaceEmbedding(
    model_name=embed_model_name,
    device="cuda" if torch.cuda.is_available() else "cpu",
    model_kwargs={"local_files_only": True}
)

Settings.llm = llm
Settings.embed_model = embed_model
```

## Indexing

```python
documents = SimpleDirectoryReader('data').load_data()
index = VectorStoreIndex.from_documents(documents)
```

## Query

```python

query_engine = index.as_query_engine()
response = query_engine.query(
    "summarize each document in a few sentences"
)
print(response)
```

```python
import pandas as pd
import inspect

def dump_module_structure(module_, filter_fn=lambda _m: True):
  rows = []
  def get_modules(m, filter_fn=lambda _m: True, level=0):
    if level > 0:
      return
    for mm in inspect.getmembers(m, filter_fn):
      rows.append(
          (mm[1].__name__, (mm[1].__doc__ or '').replace("\n", " ")[:200]))
      get_modules(mm[1], filter_fn, level+1)


  get_modules(m=module_,
              filter_fn=filter_fn)
  pd.set_option('display.max_rows', None)
  df = pd.DataFrame(rows, columns=["module", "description"])
  return df
```

```python
import llama_index.core
import llama_index.embeddings
import llama_index.llms
import llama_index.readers
import llama_index.retrievers
import llama_index.vector_stores

df = dump_module_structure(llama_index, lambda _m: inspect.ismodule(_m) or inspect.isclass(_m) or inspect.isfunction(_m))
df.style.set_table_styles([
    {'selector': 'th', 'props': [('text-align', 'left')]}
]).set_properties(**{'text-align': 'left'})
```
